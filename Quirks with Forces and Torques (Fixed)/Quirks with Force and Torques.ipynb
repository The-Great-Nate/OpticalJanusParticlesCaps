Basically, if we increase the resolution of the simulation (i.e decrease the timestep size), funky things happen.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
one_ee_four = pd.read_excel("janus_prototype_rotation_1e-4.xlsx")
one_ee_five = pd.read_excel("janus_prototype_rotation_1e-5.xlsx")
one_ee_six = pd.read_excel("janus_prototype_rotation_1e-6.xlsx")
one_ee_original = pd.read_excel("bigsim_8dip.xlsx") # Original yml file for original code except with new polarising beams.
one_ee_original_five = pd.read_excel("bigsim_8dip_1e-5.xlsx") # Original yml file for original code except with new polarising beams.

## Freely_Modify to plot things on the fly.
freely_modify = pd.read_excel("janus_prototype_rotation.xlsx")
```

```python
fig, ax = plt.subplots()
fig, ax2 = plt.subplots()
fig, ax3 = plt.subplots()
freely_modify.plot("time(s)", "Cx0(Nm)", ax = ax, color="tab:blue")
freely_modify.plot("time(s)", "Cy0(Nm)", ax = ax2, color="tab:orange")
freely_modify.plot("time(s)", "Cz0(Nm)", ax = ax3, color="tab:green")
ax.set_ylabel("Torque(Nm)", fontsize=18)
ax.set_xlabel("Time(s)", fontsize=18)
ax.tick_params(axis='both', labelsize=14)
plt.legend(fontsize=14)
ax2.set_ylabel("Torque(Nm)", fontsize=18)
ax2.set_xlabel("Time(s)", fontsize=18)
ax2.tick_params(axis='both', labelsize=14)
plt.legend(fontsize=14)
ax3.set_ylabel("Torque(Nm)", fontsize=18)
ax3.set_xlabel("Time(s)", fontsize=18)
ax3.tick_params(axis='both', labelsize=14)
plt.legend(fontsize=14)
# plt.savefig("XZ-Torque.pdf")

fig, ax = plt.subplots()
fig, ax2 = plt.subplots()
fig, ax3 = plt.subplots()
freely_modify.plot("time(s)", "Fx0(N)", ax = ax, color="tab:blue")
freely_modify.plot("time(s)", "Fy0(N)", ax = ax2, color="tab:orange")
freely_modify.plot("time(s)", "Fz0(N)", ax = ax3, color="tab:green")
ax.set_ylabel("Force(N)", fontsize=18)
ax.set_xlabel("Time(s)", fontsize=18)
ax.tick_params(axis='both', labelsize=14)
plt.legend(fontsize=14)
ax2.set_ylabel("Force(N)", fontsize=18)
ax2.set_xlabel("Time(s)", fontsize=18)
ax2.tick_params(axis='both', labelsize=14)
plt.legend(fontsize=14)
ax3.set_ylabel("Force(N)", fontsize=18)
ax3.set_xlabel("Time(s)", fontsize=18)
ax3.tick_params(axis='both', labelsize=14)
plt.legend(fontsize=14)
# plt.savefig("XZ-Torque.pdf")
```
