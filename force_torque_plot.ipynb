```python
DLSR = pd.read_excel("DiamondLeft-SIlicaRight.xlsx")
DLSR

# Extract data
time = DLSR['time(s)']
x = DLSR['x0(m)']
y = DLSR['y0(m)']
z = DLSR['z0(m)']

# Create 3D plot
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Plot with actual time values for color
sc = ax.scatter(x, y, z, c=time, cmap='plasma', s=20, alpha=0.8)

# Mark start and end points
ax.scatter(x.iloc[0], y.iloc[0], z.iloc[0], 
           c='green', s=150, marker='*', label='Start (t=0 s)')
ax.scatter(x.iloc[-1], y.iloc[-1], z.iloc[-1], 
           c='red', s=150, marker='X', label=f'End (t={time.iloc[-1]:.3f} s)')

# Labels and title
ax.set_xlabel('$x$ (m)', fontsize=12)
ax.set_ylabel('$y$ (m)', fontsize=12)
ax.set_zlabel('$z$ (m)', fontsize=12)
ax.set_title('3D Trajectory of Janus Particle\n(Color: Time in seconds)', fontsize=14)
ax.legend()

# Add colorbar with actual time values
cbar = fig.colorbar(sc, ax=ax, shrink=0.6, pad=0.1)
cbar.set_label('Time (s)', fontsize=12)

# Format colorbar ticks with actual time values
cbar.set_ticks([time.min(), time.max()])
cbar.set_ticklabels([f'{time.min():.3f}', f'{time.max():.3f}'])

plt.tight_layout()
plt.savefig('3d_trajectory_actual_time_DLSR.pdf', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

DLSR = pd.read_excel("DiamondLeft-SIlicaRight.xlsx")
DRSL = pd.read_excel("DiamondRight-SilicaLeft.xlsx")

time_dlsr = DLSR['time(s)']
x_dlsr = DLSR['x0(m)']
y_dlsr = DLSR['y0(m)']
z_dlsr = DLSR['z0(m)']

time_drsl = DRSL['time(s)']
x_drsl = DRSL['x0(m)']
y_drsl = DRSL['y0(m)']
z_drsl = DRSL['z0(m)']

fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(111, projection='3d')

sc1 = ax.scatter(x_dlsr, y_dlsr, z_dlsr, c=time_dlsr, cmap='plasma', s=15, alpha=0.7, 
                 label='Diamond Left, Silica Right (DLSR)')

ax.scatter(x_dlsr.iloc[0], y_dlsr.iloc[0], z_dlsr.iloc[0], 
           c='green', s=100, marker='*', alpha=0.9)
ax.scatter(x_dlsr.iloc[-1], y_dlsr.iloc[-1], z_dlsr.iloc[-1], 
           c='red', s=100, marker='X', alpha=0.9)

sc2 = ax.scatter(x_drsl, y_drsl, z_drsl, c=time_drsl, cmap='viridis', s=15, alpha=0.7, 
                 label='Diamond Right, Silica Left (DRSL)')

ax.scatter(x_drsl.iloc[0], y_drsl.iloc[0], z_drsl.iloc[0], 
           c='green', s=100, marker='*', alpha=0.9)
ax.scatter(x_drsl.iloc[-1], y_drsl.iloc[-1], z_drsl.iloc[-1], 
           c='red', s=100, marker='X', alpha=0.9)

ax.view_init(elev=90, azim=-90)
ax.set_xlabel('$x$ (m)', fontsize=14)
ax.set_ylabel('$y$ (m)', fontsize=14)
ax.set_zlabel('$z$ (m)', fontsize=14)
ax.set_title('Comparison of Janus Particle Trajectories\nDiamond-Silica Orientation Effects', fontsize=16)
ax.legend(fontsize=12)

cbar1 = fig.colorbar(sc1, ax=ax, shrink=0.5, pad=0.05, label='Time (s) - DLSR')
cbar2 = fig.colorbar(sc2, ax=ax, shrink=0.5, pad=0.15, label='Time (s) - DRSL')

plt.tight_layout()
plt.savefig('combined_3d_trajectories.pdf', dpi=300, bbox_inches='tight')
plt.show()
```
